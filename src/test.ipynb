_Setup_

````python
import os
from dotenv import load_dotenv
from anthropic import Anthropic
import json

load_dotenv()

claude_api_key = os.getenv("CLAUDE_API_KEY")

# client = Anthropic(base_url="http://localhost:11434", api_key="ollama")
client = Anthropic(api_key=claude_api_key)

# model = "gemma4:e4b-mlx"
model = "claude-sonnet-4-5"
max_tokens = 1000
````

_Utility Functions_

````python
def add_user_message(messages, message):
  messages.append({"role": "user", "content": message})

def add_assistant_message(messages, message):
  messages.append({"role": "assistant", "content": message})

def chat(messages, **kargs):
  params = {
    "model": model,
    "max_tokens": max_tokens,
    "messages": messages,
    **kargs
  }

  response = client.messages.create(**params)

  answer = ""
  for block in response.content:
    if (block.type == "text"):
      answer = block.text

  return answer
````

## System Prompt

````python
system = "You are a patient match teacher, Do not directly answer the student's questions. Guide them step by step to a solution with hints"

messages = []
add_user_message(messages, "What is 2+2?")
resp = chat(messages, system=system)
print (resp)
````

## Temperature

````python
messages = []
add_user_message(messages, "What is global warming, answer 1 line")
resp = chat(messages, temperature=0.7)
print(resp)
````

## Structured Data

````python
messages = []
add_user_message(messages, "Generate 3 distinct and simple aws cli commands")
add_assistant_message(messages, "Here are 3 distinct and simple aws cli commands without any comments in a single code block\n```bash")
resp = chat(messages, stop_sequences=["```"])
print(resp)
````

## Chat Exercise
